```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Extract and preparation data

```python
data_negative = pd.read_csv('data/processedNegative.csv')
data_neutral = pd.read_csv('data/processedNeutral.csv')
data_positive = pd.read_csv('data/processedPositive.csv')
```

```python
data_negative = data_negative.T
data_neutral = data_neutral.T
data_positive = data_positive.T
```

```python
print(data_negative.shape)
print(data_neutral.shape)
print(data_positive.shape)
```

```python
data_negative['sentement'] = 'negative'
data_neutral['sentement'] = 'neutral'
data_positive['sentement'] = 'positive'
```

```python
data_negative.index.name = 'sentence'
data_neutral.index.name = 'sentence'
data_positive.index.name = 'sentence'
```

```python
data_positive.head(5)
```

```python
df = pd.concat([data_negative, data_neutral, data_positive])
```

```python
import re

leak_words = ['happy', 'unhappy']

def remove_leak(text):
    text = str(text)
    for w in leak_words:
        text = re.sub(rf'\b{w}\b', '', text, flags=re.IGNORECASE)
    return text.strip()

df.index = df.index.map(remove_leak)
```

Splitting (80/20, stratified by 'sentement')

```python
train, test = train_test_split(df, test_size=0.2, stratify=df['sentement'])
```

2. **Stemming**

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

stemmer = PorterStemmer()
```

```python
def apply_stemming(text):
    words = word_tokenize(text)
    stemmed = [stemmer.stem(w) for w in words]
    return ' '.join(stemmed)
```

```python
X_train_stem = X_train.map(apply_stemming)
X_test_stem  = X_test.map(apply_stemming)
```

```python
vectorizer = CountVectorizer(binary=True)
X_train_stem_vec = vectorizer.fit_transform(X_train_stem)
X_test_stem_vec  = vectorizer.transform(X_test_stem)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_stem_vec, y_train)
preds = model.predict(X_test_stem_vec)

print("0/1:", accuracy_score(y_test, preds))
```

3. **Lemmatization**

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

def apply_lemmatization(text):
    words = word_tokenize(str(text))
    lemmas = [lemmatizer.lemmatize(w) for w in words]
    return ' '.join(lemmas)
```

```python
X_train_lemm = X_train.map(apply_lemmatization)
X_test_lemm  = X_test.map(apply_lemmatization)
```

4. **Stemming + Misspellings**

```python
from spellchecker import SpellChecker

spell = SpellChecker()

def fix_misspellings(text):
    words = str(text).split()
    fixed = []
    for w in words:
        correction = spell.correction(w)
        fixed.append(correction if correction else w)
    return ' '.join(fixed)
```

```python
X_train_spel = X_train.map(fix_misspellings)
X_train_stem_spel = X_train_spel.map(apply_stemming)
X_test_spel = X_test.map(fix_misspellings)
X_test_stem_spel  = X_test_spel.map(apply_stemming)
```

**Similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
texts = df.index.to_series().reset_index(drop=True)
texts = texts.astype(str)

vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(texts)
print("matrix shape:", tfidf.shape)
```

```python
sim = cosine_similarity(tfidf)
print("proximity matrix:", sim.shape)
```

```python
sim_clean = np.triu(sim, k=1)

N = 10
flat = sim_clean.flatten()
top_idx = flat.argsort()[-N:][::-1]

print(f"Top-{N} similarity tweets:\n")
for idx in top_idx:
    i, j = np.unravel_index(idx, sim.shape)
    score = sim[i, j]
    print(f"score = {score:.3f}")
    print(f"  [{i}] {texts.iloc[i]}")
    print(f"  [{j}] {texts.iloc[j]}")
    print()
```

**Different algorithms**

```python
X_train_spel = X_train.map(fix_misspellings)
X_test_spel  = X_test.map(fix_misspellings)
X_train_prep = X_train_spel.map(apply_stemming)
X_test_prep  = X_test_spel.map(apply_stemming)

vectorizer = TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_df=0.9, sublinear_tf=True)
X_train_vec = vectorizer.fit_transform(X_train_prep)
X_test_vec  = vectorizer.transform(X_test_prep)
print("features:", X_train_vec.shape[1])
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB, ComplementNB

models = {
    "LogisticRegression": LogisticRegression(max_iter=1000, class_weight='balanced'),
    "LinearSVC":          LinearSVC(class_weight='balanced'),
    "SGDClassifier":      SGDClassifier(max_iter=1000, class_weight='balanced'),
    "MultinomialNB":      MultinomialNB(),
    "ComplementNB":       ComplementNB(),
}

results = {}
for name, m in models.items():
    m.fit(X_train_vec, y_train)
    acc = accuracy_score(y_test, m.predict(X_test_vec))
    results[name] = acc
    print(f"{name:20s}: {acc:.4f}")
```
